# Exploring the matcher

Quick playground to look at the sample dataset, see what skills get extracted, and run the matcher on a single pair before launching the full evaluation.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.matcher import match
from src.parser import read_resume, read_job_description
from src.skill_extractor import extract_skills
```

```python
# Load one resume and one JD
resume = read_resume('../data/resumes/ml_engineer.txt')
jd = read_job_description('../data/job_descriptions/ml_engineer_jd.txt')

print('Resume preview:'); print(resume[:300]); print()
print('JD preview:');     print(jd[:300])
```

```python
# What skills does the dictionary find in each?
print('Resume skills :', extract_skills(resume))
print('JD skills     :', extract_skills(jd))
```

*Your observation here:* Are any obvious skills missing from the dictionary? Things you'd want to add if you were extending this for real recruitment use?

```python
# Run the full matcher
result = match(resume, jd)
for k, v in result.items():
    if isinstance(v, list) and len(v) > 8:
        print(f'{k}: {v[:8]}... (+{len(v)-8} more)')
    else:
        print(f'{k}: {v}')
```

```python
# Sanity check: how does the same resume score against a clearly wrong JD?
wrong_jd = read_job_description('../data/job_descriptions/frontend_dev_jd.txt')
wrong_result = match(resume, wrong_jd)
print(f"ML resume vs ML JD     : {result['match_percentage']}%")
print(f"ML resume vs Frontend JD: {wrong_result['match_percentage']}%")
```

*Your observation here:* By how much does the score drop when you switch to the wrong JD? Does it drop more from the semantic component, the TF-IDF component, or the skill-overlap component? What does that tell you about which signal is doing the most work?
